```python
import numpy as np
from numpy.linalg import svd
from sklearn import datasets
from sklearn.preprocessing import StandardScaler
```

```python
# Load data
cancer = datasets.load_breast_cancer()

# Print description
print(cancer.DESCR)
```

```python
# Scale data
scaler = StandardScaler()
data_scaled = scaler.fit_transform(cancer.data)
# print data shape
print(f"data_scaled shape: {data_scaled.shape}")

# Perform SVD
U, s, Vt = svd(data_scaled, full_matrices=False)
S = np.diag(s)

# print shapes
print(f"U shape: {U.shape}")
print(f"S shape: {S.shape}")
print(f"V shape: {Vt.T.shape}")

# Display V
display(Vt.T[:, 0])
```

```python
data_scaled_T = data_scaled.T
print(f"data_scaled_T shape: {data_scaled_T.shape}")

# Perform SVD
U, s, Vt = svd(data_scaled_T, full_matrices=False)
S = np.diag(s)

# print shapes
print(f"U shape: {U.shape}")
print(f"S shape: {S.shape}")
print(f"V shape: {Vt.T.shape}")

# Display Vt
display(U[:, 0])
```

Let A be our data matrix. The above shows that V of SVD(A) equals U of SVD(A^T), as stated in solution to Exam 1 review Q2.
